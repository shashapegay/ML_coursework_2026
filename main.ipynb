```python
from sklearn.model_selection import train_test_split
import os

from config import *
from src.data_loader import load_data
from src.preprocessing import remove_nan_columns
from src.models import get_models
from src.pipeline import create_pipeline
from src.train import train_and_save
from joblib import load
```

```python
MODEL_PATH = "models/model.joblib"
```

```python
X, y = load_data(DATA_PATH)
```

```python
X = remove_nan_columns(X)
X # Колонок, состоящих целиком из NaN нет
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=TEST_SIZE, random_state=RANDOM_STATE, stratify=y
)
```

```python
if os.path.exists(MODEL_PATH):
    model = load(MODEL_PATH)
else:
    for model in get_models():
        base_model = get_models()[model]
        model = create_pipeline(base_model, PCA_COMPONENTS)
        model = train_and_save(model, X_train, y_train, MODEL_PATH)
```

```python
y_pred = model.predict(X_test)
y_pred
```
